```python
#KNN Algorithm


import numpy as np
from collections import Counter
```

```python
def euclidean_distance(point1, point2):
    return np.sqrt(np.sum((np.array(point1) - np.array(point2))**2))
```

```python
def knn_predict(training_data, training_labels, test_point, k):
    distances = []
    for i in range(len(training_data)):
        dist = euclidean_distance(test_point, training_data[i])
        distances.append((dist, training_labels[i]))
    distances.sort(key=lambda x: x[0])
    k_nearest_labels = [label for _, label in distances[:k]]
    return Counter(k_nearest_labels).most_common(1)[0][0]
```

```python
# Format: [Study Hours per day, Attendance Percentage]
training_data = [
    [1, 40],
    [2, 50],
    [1.5, 45],
    [3, 60],
    [4, 75],
    [5, 85],
    [6, 90],
    [7, 95]
]

training_labels = [
    'Fail',
    'Fail',
    'Fail',
    'Fail',
    'Pass',
    'Pass',
    'Pass',
    'Pass'
]

# New student data
test_point = [4, 92]   # 2.5 hours study, 55% attendance
k = 3
```

```python
#Classification
prediction = knn_predict(training_data, training_labels, test_point, k)
print(prediction)
```

```python
def knn_regression(training_data, training_scores, test_point, k):
    distances = []

    for i in range(len(training_data)):
        dist = euclidean_distance(test_point, training_data[i])
        distances.append((dist, training_scores[i]))

    distances.sort(key=lambda x: x[0])

    # Take first K nearest scores
    k_nearest_scores = [score for _, score in distances[:k]]

    # Return average
    return sum(k_nearest_scores) / k
```

```python


# Final Exam Scores (out of 100)
training_scores_reg = [
    35,
    45,
    40,
    55,
    70,
    80,
    90,
    95
]

# New student
test_point_reg = [4, 92]
k_reg = 3
```

```python
#regression

predicted_score = knn_regression(training_data, training_scores_reg, test_point_reg, k_reg)

print("Predicted Score:", round(predicted_score, 2))
```

```python
def multiple_linear_regression(training_data, training_scores, test_point):

    X = np.array(training_data)
    Y = np.array(training_scores)

    # Add intercept column
    ones = np.ones((X.shape[0], 1))
    X = np.hstack((ones, X))

    # Normal Equation
    beta = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(Y)

    # Add intercept to test point
    test_point = np.array([1] + test_point)

    predicted_value = test_point.dot(beta)

    return predicted_value
```

```python
predicted_score_mlr = multiple_linear_regression(
    training_data,
    training_scores_reg[:len(training_data)], # Truncate training_scores_reg to match training_data length
    test_point_reg
)

print("Predicted Score using Multiple Linear Regression:",
      round(predicted_score_mlr, 2))
```
